```python
import sys,os,re,argparse,glob
import pandas as pd
from scipy.io import loadmat
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
base_path = "/home/tntiniak/Work/observatory_benchmark/PhysiCell/results/"
folders   = ["cell_uptake_5", "cell_uptake_10", "cell_uptake_20", "cell_uptake_40"]
```

```python
for folder in folders:
# generate csv of all voxels and their diffusion in each timestep
    dh = float(folder.split("_")[-1])

# open that simulation
    sim_path = os.path.join(base_path, folder)
    files = Path(sim_path).glob('*output*_microenvironment*.mat')
    print(sim_path)
    step = 0.01
    df_cell = pd.DataFrame(columns = ['x','y','z','vol','diff'])
    i = 0
    for file in sorted(files):
        file
        mat = loadmat(file)
        mat = mat["multiscale_microenvironment"]
        steps = [step*i] * mat.shape[1]
        df_mat = pd.DataFrame(mat.transpose(),columns = ['x','y','z','vol','diff'])
        df_mat["timestep"] = steps
        df_cell= pd.concat([df_cell,df_mat])
        i= i + 1
    df_cell.to_csv(os.path.join(sim_path, "summary.csv"))
```
