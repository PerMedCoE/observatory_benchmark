```python
import sys,os,re,argparse,glob
import pandas as pd
from scipy.io import loadmat
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
output_folder = '../results/mechanics_friction/'
```

```python
files = sorted(Path(output_folder).glob('*output*_cells.mat'))

# Create empty DataFrame
df_cell = pd.DataFrame(columns=['id','x','y','z','velocity','is_motile','migration_speed','mvx','mvy','mvz','dt',"persistence_time"])

# Assume mechanical time step is 0.1 min
mechanical_dt = 0.1

# Loop through files
for idx, file in enumerate(files):
    mat = loadmat(file)
    # Extract the specific columns (according to your indices)
    cell_data = mat['cells'][[0,1,2,3,17,50,52,57,58,59,51]]
    
    # Create a DataFrame from the data
    df_mat = pd.DataFrame(cell_data.transpose(), columns=['id','x','y','z','velocity','is_motile','migration_speed','mvx','mvy','mvz','persistence_time'])
    
    # Add the current simulation time
    df_mat['dt'] = idx * mechanical_dt
    
    # Append to the main DataFrame
    df_cell = pd.concat([df_cell, df_mat], ignore_index=True)

# Save to CSV
df_cell.to_csv(Path(output_folder) / "cell_position_time.csv", index=False)
```
