```python
import pandas as pd
from scipy.io import loadmat
from pathlib import Path

import numpy as np
```

```python
output_folder = "../results/custom_stochastic_cell_cycle/"
```

```python
df_cell = pd.DataFrame(columns=['id', 'total_volume', 'current_phase', 'dt'])

# Set time interval
interval = 10.0
i = 0

# Get list of .mat files matching pattern
files = sorted(Path(output_folder).glob('*output*_cells.mat'))

# Loop through files
for file in files:
    mat = loadmat(file)
    # Extract the required rows from 'cells' array
    cell_data = mat['cells'][[0, 4, 7,8]]
    # Transpose and create a DataFrame
    df_mat = pd.DataFrame(cell_data.T, columns=['id', 'total_volume', 'current_phase',"elapsed_time_in_phase"])
    # Add timepoint
    df_mat['dt'] = round(i * interval, 1)
    # Append to main DataFrame
    df_cell = pd.concat([df_cell, df_mat], ignore_index=True)
    i += 1

# Sort by dt and reset index
# df_cell = df_cell.sort_values(by='dt').reset_index(drop=True)

# Save to CSV
df_cell.to_csv(Path(output_folder) / "cell_volumes.csv", index=False)
df_cell
```
