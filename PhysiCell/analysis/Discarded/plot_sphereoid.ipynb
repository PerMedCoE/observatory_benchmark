```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.io import loadmat
from natsort import natsorted

# Directory containing the .mat cell files
output_dir = "/home/tntiniak/Work/physicellforbenchmarknewintegrationwithrandy/PhysiCell/spheriod"
mechanical_dt = 0.1  # time step in minutes — adjust based on your simulation settings
```

```python

# Find all _cells_physicell.mat files
files = [f for f in os.listdir(output_dir) if f.endswith("_cells_physicell.mat")]
files = natsorted([os.path.join(output_dir, f) for f in files])
```

```python
for idx, file in enumerate(files):
    mat = loadmat(file)
    
    try:
        cell_data = mat['cells'][[0,1,2,3,17,50,52,57,58,59,51]]
    except KeyError:
        print(f"Missing expected keys in {file}")
        continue
    
    df = pd.DataFrame(cell_data.transpose(), columns=[
        'id','x','y','z'
    ])
    df['time_min'] = idx * mechanical_dt

    # Plot
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(df['x'], df['y'], df['z'], s=10, c='blue', alpha=0.6)

    ax.set_xlabel("X (µm)")
    ax.set_ylabel("Y (µm)")
    ax.set_zlabel("Z (µm)")
    ax.set_title(f"Time = {df['time_min'].iloc[0]:.2f} min")

    plt.tight_layout()
    plt.show()
```
