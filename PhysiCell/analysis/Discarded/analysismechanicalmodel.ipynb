```python
import glob
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# add base path
base_path = "/home/tntiniak/Work/observatory_benchmark/PhysiCell/"
```

```python
# for folder that starts with dt
output_folder = base_path + "/results/mechanics_pushing/"
distances = pd.DataFrame()
for folder in glob.glob(output_folder + "dt*"):
    df = pd.read_csv(folder + "/cells_position_time.csv")
    # calculate distance of cells with different IDs at each timepoint
    x0 = df[df['id'] == 0]['x'].reset_index(drop=True)
    x1 = df[df['id'] == 1]['x'].reset_index(drop=True)
    id = folder.split("dt")[-1].strip("/")
    # get only first 200 timepoints
    distances[id] = (x1 - x0).abs()
distances
```

```python
# plot each columr as lineplot
# add opacity to lines
plt.figure(figsize=(10,6))

for col in distances.columns:
    # plot only first 200 timepoints
    plt.plot(distances[col][:200], label=col, alpha=0.5)
    # legend is the column that holds the id
    plt.legend(distances.columns)
    
plt.xlabel("Timepoint")
plt.ylabel("Distance between cells (micron)")
```
