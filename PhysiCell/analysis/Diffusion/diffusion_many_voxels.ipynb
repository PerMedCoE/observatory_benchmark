```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
pd.set_option('display.float_format', lambda x: '%.5f' % x)
# 1 uM = 1 umol/L = 1e-6 mol/L = 6.022e+17 #/L = 602.2 #/um^3
```

```python
file = "../code/many_diffusion_voxel_sink_20/summary.csv"
pc_df= pd.read_csv(file,index_col=0)
pc_df = pc_df.loc[pc_df.index == 13]

file = "../code/many_diffusion_voxel_sink_40/summary.csv"
pc_df_40= pd.read_csv(file,index_col=0)
pc_df_40 = pc_df_40.loc[pc_df_40.index == 13]

file = "../code/many_diffusion_voxel_sink_10/summary.csv"
pc_df_10= pd.read_csv(file,index_col=0)
pc_df_10 = pc_df_10.loc[pc_df_10.index == 13]
```

```python
file = "../code/many_diffusion_voxel_sink_40_upt/summary.csv"
pc_df_40u= pd.read_csv(file,index_col=0)
pc_df_40u = pc_df_40u.loc[pc_df_40u.index == 13]

file = "../code/many_diffusion_voxel_sink_40_extra/summary.csv"
pc_df_40e = pd.read_csv(file,index_col=0)
pc_df_40e = pc_df_40e.loc[pc_df_40e.index == 13]
```

```python
file = "../code/many_diffusion_voxel_sink_10_upt/summary.csv"
pc_df_10u= pd.read_csv(file,index_col=0)
pc_df_10u = pc_df_10u.loc[pc_df_10u.index == 13]

file = "../code/many_diffusion_voxel_sink_10_extra/summary.csv"
pc_df_10e = pd.read_csv(file,index_col=0)
pc_df_10e = pc_df_10e.loc[pc_df_10e.index == 13]
```

```python
plt.plot(pc_df_40['timestep'],pc_df_40['diff']/602.2,label = 'Physicell_40',color='red',alpha=0.6)
plt.plot(pc_df_40u['timestep'],pc_df_40u['diff']/602.2,label = 'Physicell_40_uptake',color='red',linestyle='dashed',alpha=0.6)
plt.plot(pc_df_40e['timestep'],pc_df_40e['diff']/602.2,label = 'Physicell_40_extra',color='red',linestyle='dotted',alpha=0.6)
plt.plot(pc_df['timestep'],pc_df['diff']/602.2,label = 'Physicell_20',color='green',alpha=0.6)
plt.plot(pc_df_10['timestep'],pc_df_10['diff']/602.2,label = 'Physicell_10',color='blue',alpha=0.6)
plt.plot(pc_df_10u['timestep'],pc_df_10u['diff']/602.2,label = 'Physicell_10_uptake',color='blue',linestyle='dashdot',alpha=0.6)
plt.plot(pc_df_10e['timestep'],pc_df_10e['diff']/602.2,label = 'Physicell_10_extra',color='blue',linestyle='dotted',alpha=0.6)
plt.xticks(np.arange(0, 11, step=1))
plt.yticks(np.arange(0, 11, step=1))

plt.ylabel("Concentration μM")
plt.xlabel("Time (minutes)")
plt.legend(loc="best")

plt.title("Diffusion one cell as sink : Oxygen concentration in Central Voxel")
ax_inset = plt.axes([0.25, 0.55, 0.25, 0.25])  # [left, bottom, width, height]
ax_inset.plot(pc_df_40['timestep'], pc_df_40['diff'] / 602.2, color='red', alpha=0.6)
ax_inset.plot(pc_df_40u['timestep'], pc_df_40u['diff'] / 602.2, color='red', linestyle='dashed',alpha=0.6)
ax_inset.plot(pc_df_40e['timestep'], pc_df_40e['diff'] / 602.2, color='red', linestyle='dotted',alpha=0.6)
ax_inset.plot(pc_df['timestep'], pc_df['diff'] / 602.2, color='green', alpha=0.6)
ax_inset.plot(pc_df_10['timestep'], pc_df_10['diff'] / 602.2, color='blue', alpha=0.6)
ax_inset.plot(pc_df_10u['timestep'], pc_df_10u['diff'] / 602.2, color='blue', linestyle='dashdot',alpha=0.6)
ax_inset.plot(pc_df_10e['timestep'], pc_df_10e['diff'] / 602.2, color='blue', linestyle='dotted',alpha=0.6)

# Set limits for the inset
ax_inset.set_xlim(0, 0.05)
ax_inset.set_ylim(bottom=0)  # Adjust as needed for your data

# Optional: Add labels to the inset
ax_inset.set_xlabel("Time (min)")
# ax_inset.set_ylabel("Concentration μM")
plt.savefig("./diffusion_single.png",dpi=200)
plt.show()
```

```python
plt.plot(pc_df_40u['timestep'],pc_df_40u['diff']/602.2,label = 'Physicell_40',color='red',linestyle='dashed',alpha=0.6)
plt.plot(pc_df['timestep'],pc_df['diff']/602.2,label = 'Physicell_20',color='green',alpha=0.6)
plt.plot(pc_df_10u['timestep'],pc_df_10u['diff']/602.2,label = 'Physicell_10',color='blue',linestyle='dashdot',alpha=0.6)
plt.xticks(np.arange(0, 11, step=1))
plt.yticks(np.arange(0, 11, step=1))

plt.ylabel("Concentration μM")
plt.xlabel("Time (minutes)")
plt.legend(loc="best")

plt.title("Diffusion bulk uptake: mean substrate concentration")
ax_inset = plt.axes([0.25, 0.55, 0.25, 0.25])  # [left, bottom, width, height]
ax_inset.plot(pc_df_40u['timestep'], pc_df_40u['diff'] / 602.2, color='red', linestyle='dashed',alpha=0.6)
ax_inset.plot(pc_df['timestep'], pc_df['diff'] / 602.2, color='green', alpha=0.6)
ax_inset.plot(pc_df_10u['timestep'], pc_df_10u['diff'] / 602.2, color='blue', linestyle='dashdot',alpha=0.6)

# Set limits for the inset
ax_inset.set_xlim(0, 0.05)
ax_inset.set_ylim(bottom=0)  # Adjust as needed for your data
ax_inset.set_yticks(np.arange(0, 11, step=2))
# Optional: Add labels to the inset
ax_inset.set_xlabel("Time (min)")
# ax_inset.set_ylabel("Concentration μM")
plt.savefig("./diffusion_single.png",dpi=200)
plt.show()
```
