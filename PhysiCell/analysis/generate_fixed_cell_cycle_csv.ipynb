```python
import pandas as pd
from scipy.io import loadmat
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
```

```python
output_folder = "../results/fixed_cell_cycle/"
```

```python
df_cell = pd.DataFrame(columns=['id', 'x', 'y', 'z', 'total_volume', 'current_phase', 'dt'])

# Set time interval
interval = 30
i = 0

# Get list of .mat files matching pattern
files = sorted(Path(output_folder).glob('*output*_cells.mat'))

# Loop through files
for file in files:
    mat = loadmat(file)
    # Extract the required rows from 'cells' array
    cell_data = mat['cells'][[0, 1, 2, 3, 4, 7]]
    # Transpose and create a DataFrame
    df_mat = pd.DataFrame(cell_data.T, columns=['id', 'x', 'y', 'z', 'total_volume', 'current_phase'])
    # Add timepoint
    df_mat['dt'] =i * interval
    # Append to main DataFrame
    df_cell = pd.concat([df_cell, df_mat], ignore_index=True)
    i += 1

# Sort by dt and reset index
# df_cell = df_cell.sort_values(by='dt').reset_index(drop=True)

# Save to CSV
df_cell.to_csv(Path(output_folder) / "cell_volumes.csv", index=False)
df_cell
```
