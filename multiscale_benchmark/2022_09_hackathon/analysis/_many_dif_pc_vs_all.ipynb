```python
file = "../Biodynamo/unit_test_diffusion_1k/data.csv"
bd_df= pd.read_csv(file,index_col=None,header=None,sep = " ",names = ['timestep','avg_diff'])
```

```python
file = "../Tisim/diffusion 1k cells.csv"
ts_df= pd.read_csv(file,names = ['timestep','diff'],header=0)
```

```python
file = "../Chaste/unit_test_diffusion_1k/results/TestDiffusionSmall12.dat"
ch_df= pd.read_csv(file,sep='\s+',names = ['timestep','diff','diff_cells'],header=0)
```

```python
plt.plot(bd_df['timestep'],bd_df['avg_diff'],label = 'Biodynamo',alpha=0.5,color = 'red')
plt.plot(ts_df['timestep'],ts_df['diff'],label = 'TiSim', color  = '#ffd343')
plt.plot(ch_df['timestep'],ch_df['diff_cells'],label = 'Chaste Sinks')
plt.plot(ch_df['timestep'],ch_df['diff'],label = 'Chaste', color  = 'blue')
plt.plot(dx20['dt'],dx20['diff'],label = 'PC dx = 20',color='#99FF99')
# plt.plot(result_all.index,result_all['diff'],label = 'PC dx=20',color='green')
plt.plot(dx10['dt'],dx10['diff'],label = 'PC dx=10',color='black')
plt.plot(dx5['dt'],dx5['diff'],label = 'PC dx=5',color='black')

plt.ylabel("Concentration")
plt.xlabel("Time (minutes)")
plt.legend()
plt.title("Average Concentration of all voxels")
# plt.savefig("./updated_average_concentration_of_02_of_all_voxels_1k.png",dpi=200)
plt.show()
```

```python
fig, ax = plt.subplots()

# tools = ['BM', 'CH_all','CH_sinks', 'TS', 'PC_all HR','PC_allV=1','PC_sinks HR','PC_sinksV=1']
tools = ['BM', 'CH_all','CH_sinks', 'TS', 'PC_20','PC_10','PC_5']
conc = [bd_df['avg_diff'].iloc[-1], ch_df["diff"].iloc[-1],ch_df["diff_cells"].iloc[-1], ts_df['diff'].iloc[-1], dx20['diff'].iloc[-1],dx10['diff'].iloc[-1],dx5['diff'].iloc[-1]]
# bar_labels = ['red', 'blue', 'yellow', 'green','#99FF99']
bar_colors =  ['red', 'blue','blue' ,'yellow', 'green','green','green'] 
# ['tab:red', 'tab:blue', 'tab:red', 'tab:orange']

bars = ax.bar(tools, conc, label=tools, color=bar_colors)

for bar in bars:
    print(bar.get_x())
    yval = bar.get_height()  # Get the height of the bar (i.e., the value)
    ax.text(round(bar.get_x(),4) + bar.get_width() / 2, yval,    # Position the text at the center of the bar
             f'{yval}', ha='center', va='bottom')  # Display the value above the bar

ax.set_ylabel('')
ax.set_title('Convergence of Concentration at 10 minutes')
plt.setp(ax.get_xticklabels(), rotation=30, horizontalalignment='right')
plt.show()
```
