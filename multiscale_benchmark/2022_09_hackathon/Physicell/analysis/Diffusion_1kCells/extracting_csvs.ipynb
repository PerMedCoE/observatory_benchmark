```python
import sys,os,re,glob
import pandas as pd
from scipy.io import loadmat
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image
import mat73
import numpy as np
```

```python
output_folder="../../output/diffusion_1k/test/"
# output_folder="/home/tntiniak/Work/mnt/dif_dx10/"
```

```python
files = Path(output_folder).glob('*output*_microenvironment*.mat')

step = 0.1
```

```python
i = 0
dfs = []
for file in sorted(files):
    mat = loadmat(file)        # mat = mat[]
    mat = mat["multiscale_microenvironment"] # will this copy the array?
    # mean = np.mean(mat,axis=1,dtype=np.float64)
    
    steps = [step*i] * mat.shape[1]
    # dfs.append([step*i,mean[4]])

    df_mat = pd.DataFrame(mat.transpose(),columns = ['x','y','z','vol','diff'])
    df_mat["timestep"] = steps
    dfs.append(df_mat)
    i= i + 1
    # print(i)
    # # if i==100: break

df_env = pd.concat(dfs,ignore_index=True)
# df_env = pd.DataFrame(dfs,columns=['dt','diff'])

```
