```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
file = "../../output/diffusion_1k/diffusion_3/microenv_many_diffusion.csv"
pc_df= pd.read_csv(file,index_col=None,usecols=['timestep','diff'])

```

```python
pc_cells = pc_df[pc_df['diff'] != 6022.0]
```

```python
group = pc_df.groupby(by="timestep")
result_all = group.mean()
result_all['diff'] = result_all['diff']/602.2
```

```python
group = pc_cells.groupby(by="timestep")
result_cells = group.mean()
result_cells['diff'] = result_cells['diff']/602.2
```

```python
file = "../../output/diffusion_1k/diffusion_1/cells_internalized.csv"
pc_df= pd.read_csv(file,index_col=0,header=0,usecols=['id','internalized_total_substrates','timestep'])
```

```python
plt.plot(result_all.index,result_all['diff'],label = 'Mean conc of all voxels',color='green')
plt.plot(result_cells.index,result_cells['diff'],label = 'Mean conc of sink voxels',color='red')
# plt.plot(result_all_2.index,result_all_2['diff'],label = 'Mean conc of all voxels dt=0.001',color='blue')
# plt.plot(result_cells_2.index,result_cells_2['diff'],label = 'Mean conc of sink voxels dt=0.001',color='yellow')
plt.ylabel("Concentration")
plt.xlabel("Time (minutes)")
plt.legend()
plt.title("Concentration Physicell")
# plt.savefig("./concentration.png",dpi=200)
plt.show()
```
